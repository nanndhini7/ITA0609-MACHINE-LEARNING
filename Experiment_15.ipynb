```python
from sklearn.datasets import load_iris
from sklearn.naive_bayes import GaussianNB

iris = load_iris()
X = iris.data
y = iris.target

model = GaussianNB()
model.fit(X,y)

pred = model.predict([[5.1,3.5,1.4,0.2]])

print("Predicted Flower:",
      iris.target_names[pred[0]])
```
