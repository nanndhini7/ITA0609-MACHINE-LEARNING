```python
from sklearn.naive_bayes import GaussianNB

X = [
 [12,820,5],[10,790,6],[9,760,7],
 [7,700,8],[6,680,9],[5,650,10],
 [4,620,12],[8,730,8]
]

y = [
 'Approved','Approved','Approved','Approved',
 'Rejected','Rejected','Rejected','Approved'
]

model = GaussianNB()
model.fit(X,y)

print("Predicted Loan Status:",
      model.predict([[9,770,6]])[0])
```
