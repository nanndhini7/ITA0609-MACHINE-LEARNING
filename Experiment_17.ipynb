```python
from sklearn.linear_model import LinearRegression

X = [
 [4,64,12],[4,128,16],[6,128,48],
 [6,256,50],[8,128,64],[8,256,64],
 [12,256,108],[12,512,108]
]

y = [12000,15000,18000,22000,
     25000,30000,38000,45000]

model = LinearRegression()
model.fit(X,y)

print("Predicted Mobile Price:",
      model.predict([[8,256,64]])[0])
```
