```python
from sklearn.linear_model import LinearRegression

X = [
 [10,2,120],[15,3,150],[20,4,180],
 [25,5,210],[30,6,240],[35,7,260],
 [40,8,280],[45,9,300]
]

y = [3.5,4.5,5.8,7.0,8.2,9.5,10.8,12.0]

model = LinearRegression()
model.fit(X,y)

print("Predicted Future Sales:",
      model.predict([[38,7,270]])[0], "Lakhs")
```
