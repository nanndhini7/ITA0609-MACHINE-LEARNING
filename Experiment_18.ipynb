```python
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron

iris = load_iris()
X = iris.data
y = iris.target

model = Perceptron(random_state=42, max_iter=2000)
model.fit(X,y)

pred = model.predict([[5.1,3.5,1.4,0.2]])

print("Predicted Flower:",
      iris.target_names[pred[0]])
```
