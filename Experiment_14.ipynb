```python
from sklearn.linear_model import LinearRegression

X = [
 [800,2,15],[1000,2,12],[1200,3,10],
 [1400,3,8],[1600,4,6],[1800,4,5],
 [2000,5,4],[2200,5,3]
]

y = [35,42,52,63,75,88,102,118]

model = LinearRegression()
model.fit(X,y)

print("Predicted House Price:",
      model.predict([[1700,4,5]])[0], "Lakhs")
```
