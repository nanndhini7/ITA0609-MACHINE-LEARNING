```python
from sklearn.tree import DecisionTreeClassifier

X = [
 [12,10,1],[10,8,2],[8,6,2],
 [6,4,3],[4,2,4],[15,12,1],
 [7,5,3],[5,3,4]
]

y = [
 'Excellent','Good','Good','Average',
 'Poor','Excellent','Average','Poor'
]

model = DecisionTreeClassifier(criterion='entropy')
model.fit(X,y)

print("Predicted Credit Score:",
      model.predict([[9,7,2]])[0])
```
