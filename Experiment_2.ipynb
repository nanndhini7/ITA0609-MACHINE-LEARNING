```python
from itertools import product

data = [
 (['Sunny','Warm','Normal','Strong'],'Yes'),
 (['Sunny','Warm','High','Strong'],'Yes'),
 (['Rainy','Cold','High','Strong'],'No'),
 (['Sunny','Warm','High','Weak'],'Yes')
]

domains = [
 ['Sunny','Rainy'],
 ['Warm','Cold'],
 ['Normal','High'],
 ['Strong','Weak']
]

all_h = list(product(*[
    d + ['?'] for d in domains
]))

def covers(h, x):
    return all(a == '?' or a == b for a,b in zip(h,x))

G = [h for h in all_h]
S = []

for x, label in data:
    if label == 'Yes':
        G = [h for h in G if covers(h,x)]
        S = [h for h in all_h if covers(h,x)] if not S else [
            h for h in S if covers(h,x)
        ]
    else:
        G = [h for h in G if not covers(h,x)]
        S = [h for h in S if not covers(h,x)]

print("Consistent hypotheses:", len(G))
print("General boundary:", G)
```
