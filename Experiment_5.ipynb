```python
from sklearn.neighbors import KNeighborsClassifier

X = [
 [150,7.5], [160,7.8], [145,7.3],
 [120,5.5], [125,5.8], [130,6.0]
]

y = ['Apple','Apple','Apple',
     'Orange','Orange','Orange']

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X,y)

print("Predicted Class:", model.predict([[152,7.5]])[0])
```
