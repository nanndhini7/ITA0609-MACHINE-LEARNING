```python
from sklearn.linear_model import LinearRegression

X = [
 [1,12000,1200],[2,18000,1200],
 [3,25000,1300],[4,35000,1400],
 [5,45000,1500],[6,55000,1500],
 [7,65000,1600],[8,75000,1600]
]

y = [8.5,8.0,7.4,6.8,6.1,5.5,5.0,4.5]

model = LinearRegression()
model.fit(X,y)

print("Predicted Car Price:",
      model.predict([[4,38000,1400]])[0])
```
