```python
from sklearn.linear_model import LogisticRegression

X = [
 [9.2,90], [8.8,85], [7.2,72],
 [6.8,68], [9.0,91], [7.0,70],
 [8.7,86], [6.5,65]
]

y = [1,1,0,0,1,0,1,0]

model = LogisticRegression()
model.fit(X,y)

print("Predicted Class:", model.predict([[8.9,88]])[0])
print("Training Accuracy:", model.score(X,y))
```
